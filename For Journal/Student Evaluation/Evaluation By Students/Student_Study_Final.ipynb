```python
import pandas as pd
import numpy as np
from scipy import stats

# =========================
# 1. Load your file
# =========================
file_path = "/content/Student Evaluation Final.xlsx"  # change if needed
df = pd.read_excel(file_path)

# Clean column names
df.columns = df.columns.str.strip()

# =========================
# 2. Define dimensions
# =========================
dimensions = [
    "JobReadiness",
    "Personalization",
    "Clarity",
    "Motivation",
    "Usefulness",
    "Overall"
]

# =========================
# 3. Function: Cohen's d (paired)
# =========================
def cohens_d_paired(x, y):
    diff = x - y
    return diff.mean() / diff.std(ddof=1)

# =========================
# 4. Compute statistics
# =========================
results = []

for dim in dimensions:
    our = df[f"Our_{dim}"]
    base = df[f"Base_{dim}"]

    # Means
    mean_our = our.mean()
    mean_base = base.mean()

    # Delta
    delta = mean_our - mean_base

    # Paired t-test
    t_stat, p_value = stats.ttest_rel(our, base)

    # Cohen's d
    d = cohens_d_paired(our, base)

    results.append({
        "Dimension": dim,
        "Mean_Our": round(mean_our, 3),
        "Mean_Base": round(mean_base, 3),
        "Delta": round(delta, 3),
        "t_stat": round(t_stat, 3),
        "p_value": round(p_value, 5),
        "Cohens_d": round(d, 3)
    })

# =========================
# 5. Convert to DataFrame
# =========================
results_df = pd.DataFrame(results)

# =========================
# 6. Display results
# =========================
print(results_df)

# Optional: save to Excel
results_df.to_excel("statistical_results.xlsx", index=False)
```
